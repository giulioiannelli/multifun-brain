## Group by (contrast, processing, band)

```python
if results is None:
    by_triplet = {}
else:
    by_triplet = {}
    for r in results:
        # label looks like 'patient/sub-00246757/co2_bpfBOLD/s4'
        parts = r.label.split('/')
        subject, proc_tag, band = parts[1], parts[2], parts[3]
        by_triplet.setdefault((proc_tag, band), {})[subject] = r
    print(f'{len(by_triplet)} (proc, band) groups, each with up to 6 subjects.')
```
